```python
import tkinter as tk
import random

# =========================
# НАСТРОЙКИ
# =========================

WIDTH = 400
HEIGHT = 400
CELL = 20
SPEED = 120

# =========================
# ОКНО
# =========================

root = tk.Tk()
root.title("Snake")
root.resizable(False, False)

canvas = tk.Canvas(
    root,
    width=WIDTH,
    height=HEIGHT,
    bg="black"
)

canvas.pack()

# =========================
# ПЕРЕМЕННЫЕ ИГРЫ
# =========================

snake = []
direction = "Right"
food = None
score = 0
game_running = True


# =========================
# НАЧАЛО / ПЕРЕЗАПУСК
# =========================

def start_game():
    global snake, direction, food, score, game_running

    snake = [
        (200, 200),
        (180, 200),
        (160, 200)
    ]

    direction = "Right"
    score = 0
    game_running = True

    create_food()
    draw()

    root.after(SPEED, game)


# =========================
# СОЗДАНИЕ ЕДЫ
# =========================

def create_food():
    global food

    while True:
        new_food = (
            random.randrange(0, WIDTH, CELL),
            random.randrange(0, HEIGHT, CELL)
        )

        if new_food not in snake:
            food = new_food
            break


# =========================
# УПРАВЛЕНИЕ
# =========================

def change_direction(event):
    global direction

    if event.keysym == "Up" and direction != "Down":
        direction = "Up"

    elif event.keysym == "Down" and direction != "Up":
        direction = "Down"

    elif event.keysym == "Left" and direction != "Right":
        direction = "Left"

    elif event.keysym == "Right" and direction != "Left":
        direction = "Right"


# =========================
# ОСНОВНАЯ ИГРА
# =========================

def game():
    global snake, food, score, game_running

    if not game_running:
        return

    # Координаты головы
    head_x, head_y = snake[0]

    # Движение
    if direction == "Up":
        head_y -= CELL

    elif direction == "Down":
        head_y += CELL

    elif direction == "Left":
        head_x -= CELL

    elif direction == "Right":
        head_x += CELL

    # =========================
    # ПРОХОД СКВОЗЬ СТЕНЫ
    # =========================

    if head_x < 0:
        head_x = WIDTH - CELL

    elif head_x >= WIDTH:
        head_x = 0

    if head_y < 0:
        head_y = HEIGHT - CELL

    elif head_y >= HEIGHT:
        head_y = 0

    new_head = (head_x, head_y)

    # =========================
    # СТОЛКНОВЕНИЕ С СОБОЙ
    # =========================

    if new_head in snake:
        game_over()
        return

    # Добавляем новую голову
    snake.insert(0, new_head)

    # =========================
    # ПРОВЕРКА ЕДЫ
    # =========================

    if new_head == food:

        score += 1

        create_food()

    else:
        # Если еду не съели,
        # убираем хвост
        snake.pop()

    # Рисуем игру
    draw()

    # Следующий шаг
    root.after(SPEED, game)


# =========================
# РИСОВАНИЕ
# =========================

def draw():
    canvas.delete("all")

    # =========================
    # СЕТКА
    # =========================

    for x in range(0, WIDTH, CELL):
        canvas.create_line(
            x, 0,
            x, HEIGHT,
            fill="#111111"
        )

    for y in range(0, HEIGHT, CELL):
        canvas.create_line(
            0, y,
            WIDTH, y,
            fill="#111111"
        )

    # =========================
    # ЗМЕЙКА
    # =========================

    for i, (x, y) in enumerate(snake):

        if i == 0:
            # Голова
            canvas.create_rectangle(
                x,
                y,
                x + CELL,
                y + CELL,
                fill="lime",
                outline="white"
            )

        else:
            # Тело
            canvas.create_rectangle(
                x,
                y,
                x + CELL,
                y + CELL,
                fill="green",
                outline="black"
            )

    # =========================
    # ЕДА
    # =========================

    if food is not None:
        x, y = food

        canvas.create_oval(
            x + 2,
            y + 2,
            x + CELL - 2,
            y + CELL - 2,
            fill="red",
            outline="white"
        )

    # =========================
    # СЧЁТ
    # =========================

    canvas.create_text(
        50,
        10,
        text=f"Score: {score}",
        fill="white",
        font=("Arial", 12),
        anchor="nw"
    )


# =========================
# GAME OVER
# =========================

def game_over():
    global game_running

    game_running = False

    canvas.create_rectangle(
        70,
        140,
        330,
        260,
        fill="black",
        outline="red",
        width=2
    )

    canvas.create_text(
        WIDTH // 2,
        175,
        text="GAME OVER",
        fill="red",
        font=("Arial", 30, "bold")
    )

    canvas.create_text(
        WIDTH // 2,
        220,
        text=f"Score: {score}",
        fill="white",
        font=("Arial", 16)
    )

    canvas.create_text(
        WIDTH // 2,
        245,
        text="Press SPACE to restart",
        fill="white",
        font=("Arial", 11)
    )


# =========================
# КЛАВИАТУРА
# =========================

root.bind("<Up>", change_direction)
root.bind("<Down>", change_direction)
root.bind("<Left>", change_direction)
root.bind("<Right>", change_direction)

# SPACE = перезапуск
root.bind("<space>", lambda event: start_game())


# =========================
# ЗАПУСК
# =========================

start_game()

root.mainloop()

```
